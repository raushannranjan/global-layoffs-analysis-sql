# Crop Yield Analysis using Environmental Factors
 This project analyzes how environmental factors such as rainfall, temperature, and pesticide usage impact crop yield using Python.

## Data Loading
The dataset is loaded using pandas for analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('yield_df.csv')
df.head()

df.columns

df = df.drop(columns=['Unnamed: 0'])
```

## Rainfall vs Yield Analysis
This plot examines the relationship between rainfall and crop yield.


```python
plt.scatter(df['average_rain_fall_mm_per_year'], df['hg/ha_yield'])
plt.xlabel("Rainfall")
plt.ylabel("Yield")
plt.title("Rainfall vs Yield")
plt.show()
```

### Observation : No strong linear relationship is observed between rainfall and yield. Yield appears higher in moderate rainfall ranges, while very high rainfall does not consistently increase productivity.

## Temperature vs Yield Analysis

```python
plt.scatter(df['avg_temp'], df['hg/ha_yield'])
plt.xlabel("Temperature")
plt.ylabel("Yield")
plt.title("Temperature vs Yield")
plt.show()
```

### Observation : No clear pattern is observed between temperature and yield. Yield values are distributed across all temperature ranges, indicating temperature alone is not a dominant factor.

## Correlation Analysis
Correlation matrix is used to quantify relationships between variables.

```python
df[['hg/ha_yield','average_rain_fall_mm_per_year','avg_temp','pesticides_tonnes']].corr()

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

### Observation : No variable shows strong correlation with yield. This suggests that crop productivity depends on multiple interacting environmental factors rather than a single variable.
